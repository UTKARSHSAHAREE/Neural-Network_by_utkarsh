# Task: Add Agency — Create an Autonomous Goal-Oriented Agent

```python
class AgenticAI:
    def __init__(self,goal="Assist user effectively"):
        self.goal=goal
        self.memory = []
        
    def perceive (self,user_input):
        self.memory.append(user_input)
        return f"Recieved input: {user_input}"
        
    def decide(self):
        if "weather" in self.memory[-1]:
            return "would you like me to give 3 day forecast?"
        elif "time" in self.memory:
            return "let me check the system clock.."
        else :
            return "I am aligning my next step with your goal."
    def act(self):
        decision = self.decide()
        print(f"🤖 Agent Action: {decision}")
```

```python
agent = AgenticAI()
agent.perceive("Tell me the weather")
agent.act()
agent.perceive("What time is it?")
agent.act()
```

```python
class Assistance:
    def __init__(self):
        self.memory={}
    def respond(self,user_input):
        if "remind" in user_input.lower():
            task=user_input.split("remind me to")[-1].strip()
            self.memory["task"]=task
        elif "what's my task" in user_input.lower():
            return f"your current reminder is :{self.memory.get("task","none")}"
        else:
            return "I can store and recall task!"
```

```python
assistance = Assistance()
print(assistance.respond("Remined me to check the reports"))
print(assistance.respond("what's my task"))
```

```python
class Assistant:
    def __init__(self):
        self.memory = {}
    
    def respond(self, user_input):
        if "remind" in user_input.lower():
            task = user_input.split("remind me to")[-1].strip()
            self.memory["task"] = task
            return f"Got it! I’ll remind you to {task}."
        elif "what’s my task" in user_input.lower():
            return f"Your current reminder is: {self.memory.get('task', 'none')}."
        else:
            return "I can store and recall tasks!"
```

```python
assistant = Assistant()
print(assistant.respond("Remind me to check the reports"))
print(assistant.respond("What’s my task?"))
```

# **Agentic AI – Autonomous Behavior**

```python
class Agent:
    def __init__(self,goal):
        self.goal = goal
        self.steps=[]
    def perceive(self,environment):
        self.steps.append(f"Observed:{environment}")
    def decide(self):
        if "email" in self.steps[-1]:
            action = "draft a professional response"
        elif "data" in self.steps[-1]:
            action = "analyze trends"
        else:
            action = "ask clarifying Question"
        self.steps.append(f"Action:{action}")
        return action
    def act(self):
        action = self.decide()
        print(f"Acting:{action} towards goal'{self.goal}'")
```

```python
agent = Agent(goal="Assist in work automation")
agent.perceive("incoming email from manager")
agent.act()
agent.perceive("new dataset uploaded")
agent.act()
```

# **creating Base Agents**

```python
class BaseAgent:
    def __init__(self, name, autonomy_level):
        self.name = name
        self.autonomy_level = autonomy_level  # 1=manual, 2=semi-auto, 3=fully autonomous
        self.memory = []
    
    def perceive(self, task):
        self.memory.append(task)
        print(f"👀 {self.name} perceives: {task}")
    
    def decide(self):
        if self.autonomy_level == 1:
            return "Waiting for human input..."
        elif self.autonomy_level == 2:
            return "Suggesting next action..."
        else:
            return "Acting independently."
    
    def act(self, action):
        print(f"🤖 {self.name} performs action: {action}")
```

```python
researcher = BaseAgent("Researcher Agent", autonomy_level=3)
planner = BaseAgent("Planner Agent", autonomy_level=2)
executor = BaseAgent("Executor Agent", autonomy_level=1)
 
def multi_agent_workflow(goal):
    print(f"\n🎯 Goal: {goal}\n")
    researcher.perceive("Gather ethical AI principles")
    researcher.act("Summarize global frameworks (EU AI Act, OECD, etc.)")
    
    planner.perceive("Organize content flow")
    planner.act("Suggest structure: Introduction, Principles, Case Studies, Guardrails")
    
    executor.perceive("Prepare slides based on plan")
    executor.act("Create PowerPoint visuals under supervision")
    
    print("\n✅ Multi-Agent Workflow Complete.")
```

```python
multi_agent_workflow("Create AI Ethics Presentation")
```
